```python
from ultralytics import YOLO
```

```python
# load the trained model
model = YOLO('runs/detect/train/weights/best.pt')
```

```python
# Predict the validation dataset
results = model.predict(source="./datasets/val/images")
```

```python
mean_predict_time = sum(result.speed['inference'] for result in results) / len(results)
print(f"Mean Prediction Time: {mean_predict_time:.2f} ms")
```

```python
En = 1012
Es = 1010
Fr = 1000

# Assuming you have a list of ground truth labels corresponding to the validation dataset
ground_truth_labels = []   # Replace with the actual ground truth labels
for i in range(En):
    ground_truth_labels.append(0)  # 0 for En
for i in range(Es):
    ground_truth_labels.append(1)  # 1 for Es
for i in range(Fr):
    ground_truth_labels.append(2)  # 2 for Fr


# Ensure the length of results matches the ground truth labels
assert len(results) == len(ground_truth_labels), "Mismatch between predictions and ground truth labels"

# Calculate accuracy
correct_predictions = 0
for result, ground_truth in zip(results, ground_truth_labels):
    if len(result.boxes.cls) > 0:  # Check if there are predictions
        predicted_class = result.boxes.cls[0].item()  # Get the predicted class
        if predicted_class == ground_truth:  # Compare with ground truth
            correct_predictions += 1
print(f"Correct Predictions: {correct_predictions}")
accuracy = correct_predictions / len(ground_truth_labels)
print(f"Accuracy: {accuracy:.3f}")
```
